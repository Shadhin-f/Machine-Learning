```python
#v4
import cv2
import mediapipe as mp
import os
import numpy as np

# === Setup ===
input_folder = r'C:\Users\User\Downloads\Skin-disease-dataset\mixed-small'
output_folder_skin = r'C:\Users\User\Downloads\Skin-disease-dataset\output_skin_combined'
os.makedirs(output_folder_skin, exist_ok=True)

# === Initialize MediaPipe ===
mp_selfie = mp.solutions.selfie_segmentation
segmentor = mp_selfie.SelfieSegmentation(model_selection=1)

def get_skin_mask_hsv(image):
    """Skin detection using HSV range."""
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    lower = np.array([0, 20, 70], dtype=np.uint8)
    upper = np.array([20, 255, 255], dtype=np.uint8)
    skin_mask = cv2.inRange(hsv, lower, upper)
    return skin_mask

def get_skin_mask_ycrcb(image):
    """Alternative skin detection using YCrCb."""
    ycrcb = cv2.cvtColor(image, cv2.COLOR_BGR2YCrCb)
    lower = np.array([0, 133, 77], dtype=np.uint8)
    upper = np.array([255, 173, 127], dtype=np.uint8)
    return cv2.inRange(ycrcb, lower, upper)

# === Process ===
total, saved = 0, 0
for img_name in os.listdir(input_folder):
    if not img_name.lower().endswith(('.jpg', '.jpeg', '.png')):
        continue

    total += 1
    img_path = os.path.join(input_folder, img_name)
    image = cv2.imread(img_path)
    if image is None:
        print(f"[!] Skipped: {img_name}")
        continue

    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    seg_result = segmentor.process(image_rgb)

    # Segmentation mask
    seg_mask = (seg_result.segmentation_mask > 0.3).astype(np.uint8)

    # Skin masks
    hsv_mask = get_skin_mask_hsv(image)
    ycrcb_mask = get_skin_mask_ycrcb(image)

    # Combine all masks
    combined_mask = cv2.bitwise_and(seg_mask * 255, hsv_mask)
    combined_mask = cv2.bitwise_and(combined_mask, ycrcb_mask)

    # Postprocess: morphological filtering
    kernel = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))
    combined_mask = cv2.morphologyEx(combined_mask, cv2.MORPH_OPEN, kernel, iterations=1)
    combined_mask = cv2.medianBlur(combined_mask, 5)

    # Apply to image
    result = cv2.bitwise_and(image, image, mask=combined_mask)

    # Optional: replace background with white
    bg_white = np.ones_like(image, dtype=np.uint8) * 255
    output = np.where(combined_mask[:, :, None] == 0, bg_white, result)

    # Check if output has meaningful content
    if np.count_nonzero(combined_mask) > 1000:
        out_path = os.path.join(output_folder_skin, f"{os.path.splitext(img_name)[0]}_skin.jpg")
        cv2.imwrite(out_path, output)
        print(f"[✓] {img_name} → Saved skin-only version")
        saved += 1
    else:
        print(f"[✗] {img_name} → Not enough skin detected")

print(f"\n✅ Done: {saved}/{total} images saved with improved skin detection.")
```

```python
#v5

import cv2
import mediapipe as mp
import os
import numpy as np

# === Setup ===
input_folder = r'C:\Users\User\Downloads\Skin-disease-dataset\mixed'
output_folder = r'C:\Users\User\Downloads\Skin-disease-dataset\output_skin_clean_v5'
os.makedirs(output_folder, exist_ok=True)

# === Initialize MediaPipe Selfie Segmentation ===
mp_selfie = mp.solutions.selfie_segmentation
segmentor = mp_selfie.SelfieSegmentation(model_selection=1)

# === Process Images ===
total, saved = 0, 0

for img_name in os.listdir(input_folder):
    if not img_name.lower().endswith(('.jpg', '.jpeg', '.png')):
        continue

    total += 1
    img_path = os.path.join(input_folder, img_name)
    image = cv2.imread(img_path)
    if image is None:
        print(f"[!] Skipped unreadable: {img_name}")
        continue

    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    results = segmentor.process(image_rgb)

    # Segmentation mask: person vs background
    mask = results.segmentation_mask
    condition = mask > 0.3  # 0.5 is default, lower allows more foreground

    # Remove background (make it white)
    fg_image = np.where(condition[:, :, None], image, 255)

    # Save only if enough foreground exists
    if np.count_nonzero(condition) > 1000:
        out_path = os.path.join(output_folder, f"{os.path.splitext(img_name)[0]}_foreground.jpg")
        cv2.imwrite(out_path, fg_image)
        print(f"[✓] {img_name} → Saved skin region")
        saved += 1
    else:
        print(f"[✗] {img_name} → No person detected")

print(f"\n✅ Finished: {saved}/{total} images processed.")
```
